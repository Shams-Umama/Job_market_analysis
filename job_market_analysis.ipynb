```python
bad = ~df["Exp years"].str.contains("Yrs", na=False) & (df["Exp years"] != "Not specified")

df.loc[bad, "Skills"] = df.loc[bad, "Exp years"] + "," + df.loc[bad, "Skills"]
df.loc[bad, "Exp years"] = "Not specified"

print(bad.sum())
```

```python
df["Category"] = df["Skills"].str.split(",").str[0].str.strip()
df["Category"].value_counts().head(10)
```

```python
df.info()
```

```python
df["City"] = df["Location"].str.split(",").str[-2].str.strip()
df["City"].value_counts()
```

```python
df["Company name"] = df["Company name"].str.strip(" -")
df["Company name"].head()
```

```python
df[["Job title", "Company name", "City", "Category", "Exp years"]].head(10)
```

```python
df["Category"].value_counts()
```

```python
tags = ["males_only", "males_preferred", "females_only", "females_preferred"]

def get_category(skills):
    for item in skills.split(","):
        item = item.strip()
        if item not in tags:
            return item

def get_gender(skills):
    for item in skills.split(","):
        item = item.strip()
        if item in tags:
            return item
    return "not_mentioned"

df["Category"] = df["Skills"].apply(get_category)
df["Gender pref"] = df["Skills"].apply(get_gender)
```

```python
import numpy as np

years = df["Exp years"].str.extract(r"(\d+)\s*(?:-|to)?\s*(\d+)?")
df["Exp min"] = pd.to_numeric(years[0])
df["Exp max"] = pd.to_numeric(years[1])
```

```python
df[["Exp years", "Exp min", "Exp max"]].sample(10, random_state=1)
```

```python
df[df["Exp max"].isna() & df["Exp min"].notna()]["Exp years"].value_counts()
```

```python
both_missing = (df["Exp level"] == "Not specified") & (df["Exp years"] == "Not specified")
print(both_missing.sum())
```

```python
bins = [-1, 2, 5, 9, 100]
labels = ["0-2 yrs", "3-5 yrs", "6-9 yrs", "10+ yrs"]

df["Exp band"] = pd.cut(df["Exp min"], bins=bins, labels=labels).astype(object)
df["Exp band"] = df["Exp band"].fillna("Not specified")

df["Exp band"].value_counts()
```

```python
df = df.rename(columns={
    "Job title": "job_title",
    "Company name": "company",
    "Location": "location",
    "Job type": "job_type",
    "Exp level": "exp_level",
    "Exp years": "exp_years",
    "Skills": "skills",
    "Category": "category",
    "City": "city",
    "Gender pref": "gender_pref",
    "Exp min": "exp_min",
    "Exp max": "exp_max",
    "Exp band": "exp_band"
})
df.columns
```

```python
import sqlite3

conn = sqlite3.connect("jobs.db")
df.to_sql("jobs", conn, if_exists="replace", index=False)

def q(sql):
    return pd.read_sql(sql, conn)
```

```python
q("SELECT COUNT(*) AS total_jobs FROM jobs")
```

```python
q("SELECT job_title, company, city, category, exp_band FROM jobs LIMIT 5")
```

```python
q("""
SELECT category, COUNT(*) AS jobs
FROM jobs
GROUP BY category
ORDER BY jobs DESC
LIMIT 10
""")
```

```python
q("""
SELECT category, COUNT(*) AS fresher_jobs
FROM jobs
WHERE exp_band = '0-2 yrs'
GROUP BY category
ORDER BY fresher_jobs DESC
LIMIT 10
""")
```

```python
q("""
SELECT city, COUNT(*) AS jobs,
       ROUND(100.0 * COUNT(*) / (SELECT COUNT(*) FROM jobs), 1) AS pct
FROM jobs
GROUP BY city
ORDER BY jobs DESC
LIMIT 8
""")
```

```python
q("""
SELECT category, COUNT(*) AS total_jobs,
       ROUND(100.0 * SUM(CASE WHEN exp_band = '0-2 yrs' THEN 1 ELSE 0 END) / COUNT(*), 1) AS fresher_pct
FROM jobs
WHERE exp_band != 'Not specified'
GROUP BY category
HAVING COUNT(*) >= 20
ORDER BY fresher_pct DESC
""")
```

```python
q("""
SELECT company, COUNT(*) AS jobs
FROM jobs
WHERE company NOT LIKE '%Confidential%'
GROUP BY company
ORDER BY jobs DESC
LIMIT 10
""")
```

```python
q("""
SELECT category, COUNT(*) AS jobs, ROUND(AVG(exp_min), 1) AS avg_min_years
FROM jobs
WHERE exp_min IS NOT NULL
GROUP BY category
HAVING COUNT(*) >= 20
ORDER BY avg_min_years DESC
""")
```

```python
q("""
SELECT gender_pref, COUNT(*) AS jobs,
       ROUND(100.0 * COUNT(*) / (SELECT COUNT(*) FROM jobs), 1) AS pct
FROM jobs
GROUP BY gender_pref
ORDER BY jobs DESC
""")
```
